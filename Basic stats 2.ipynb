```python
from scipy import stats
import pandas as pd
import numpy as np
```

```python
#Data
data = {'Print-heads':[1.13,1.55,1.43,0.92,1.25,1.36,1.32,0.85,1.07,1.48,1.20,1.33,1.18,1.22,1.29]}
```

```python
sampled_data = pd.DataFrame(data)
```

```python
sampled_data
```

```python
sampled_data.mean()
```

```python
sampled_data.std()
```

```python
#a.Building 99% confidence interval using sample standard deviation:
sampled_data_ci=stats.t.interval(0.99,15,
                                  loc = sampled_data.mean(),
                                  scale = sampled_data.std())
print(sampled_data_ci)
print( 'Mean at 99% confidence interval is:', np.round(sampled_data_ci, 4))
```

The main reason behind using the t_distribution was the sigma /standard deviation is unknown to this.using t_distribution we can also build the 99% confidence interval using the sample standars deviation.to build this confidence interval,we use sample data mean,standard deviation and thet_distribution.
